## Q11 Simple Calculator

```python
num1 = float(input("Enter first number: "))
num2 = float(input("Enter second number: "))
op   = input("Enter operator (+, -, *, /): ")

if op == '+':
    print("Result:", num1 + num2)
elif op == '-':
    print("Result:", num1 - num2)
elif op == '*':
    print("Result:", num1 * num2)
elif op == '/':
    if num2 != 0:
        print("Result:", num1 / num2)
    else:
        print("Error: Division by zero!")
else:
    print("Invalid operator")
```

## Q26 Traffic Fine System

```python
speed     = int(input("Enter speed (km/h): "))
emergency = input("Emergency vehicle? (yes/no): ").lower()
weather   = input("Is it raining? (yes/no): ").lower()

if emergency == "yes":
    print("No fine – Emergency vehicle")
elif speed <= 60:
    print("No fine")
elif 61 <= speed <= 80:
    fine = 500
    if weather == "yes":
        fine = fine * 1.5
    print(f"Fine: Rs. {fine}")
elif 81 <= speed <= 100:
    fine = 2000
    if weather == "yes":
        fine = fine * 1.5
    print(f"Fine: Rs. {fine}")
else:
    print("License Suspended")
```

## Q27 Company Candidate Filter

```python
experience      = int(input("Years of experience: "))
skill_score     = int(input("Skill score (0-100): "))
test_score      = int(input("Test score (0-100): "))
criminal_record = input("Criminal record? (yes/no): ").lower()
job_switches    = int(input("Job switches in last 2 years: "))

if criminal_record == "yes":
    print("Rejected – Criminal record found")
elif job_switches > 3:
    print("Flagged as Unstable – Too many job switches")
elif experience >= 5 and skill_score >= 80:
    print("Shortlisted")
elif 2 <= experience <= 4 and skill_score >= 70:
    print("Called for Interview")
elif experience == 0 and test_score >= 85:
    print("Fresher Shortlisted – High test score")
else:
    print("Not Selected")
```

## Q28 Passenger Clearance

## Q30 Suspicious Banking Transaction Detector

```python
amount          = float(input("Transaction amount: "))
location_match  = input("Location match? (yes/no): ").lower()
time_of_day     = input("Time (day/night): ").lower()
prev_count      = int(input("Transactions in last hour: "))
device_verified = input("Device verified? (yes/no): ").lower()

if prev_count > 10:
    print("Account Frozen – Too many transactions in 1 hour")
elif location_match == "no" and amount > 100000:
    print("Fraud Alert – Location mismatch with high amount")
elif time_of_day == "night" and device_verified == "no":
    print("High Risk – Unverified device at night")
elif amount > 100000:
    print("Suspicious – Large amount transaction")
else:
    print("Normal Transaction")
```

## Q31 Police Risk Level Classifier

```python
criminal_record          = input("Criminal record? (yes/no): ").lower()
number_of_cases          = int(input("Number of cases: "))
travel_history_flagged   = input("Travel history flagged? (yes/no): ").lower()
financial_activity_score = int(input("Financial activity score (0-100): "))
last_activity_days       = int(input("Days since last activity: "))

if criminal_record == "yes" and number_of_cases > 3:
    print("High Risk")
elif travel_history_flagged == "yes" and financial_activity_score > 70:
    print("High Risk – Flagged Travel + Suspicious Financials")
elif last_activity_days > 365:
    print("Monitor – No recent activity for over a year")
elif criminal_record == "no" and financial_activity_score <= 70:
    print("Low Risk")
else:
    print("Medium Risk")
```
